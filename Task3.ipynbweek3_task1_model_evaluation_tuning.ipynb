# **Step 2: Written Explanation — Why Accuracy Is Misleading (Markdown Cell)**
## Why Accuracy Can Be Misleading

If the dataset is imbalanced (e.g., 90% of the logs are in one class and 10% in another),
the model may always predict the majority class and still show high accuracy,
without actually learning anything. For example, if 90% of the logs are "not survived" and
the model always predicts "not survived," its accuracy will be 90%, but it will never
correctly identify any survivors. This is why Precision, Recall, and F1-score are important—they indicate how accurately the model is identifying the minority class (here "survived"), not just overall accuracy.

# Step 3: Hyperparameter Tuning — GridSearchCV


Now let's systematically test the model settings (hyperparameters) to find the best combination, rather than guessing.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

# Jo parameters test karne hain, unki grid banao
param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs']
}

grid_search = GridSearchCV(
    LogisticRegression(max_iter=1000),
    param_grid,
    cv=5,
    scoring='f1'
)

grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best F1 Score (cross-validation):", grid_search.best_score_)
```

# Step 4: Extract new predictions from the best model

```python
best_model = grid_search.best_estimator_

y_pred_tuned = best_model.predict(X_test)

from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score

accuracy_tuned = accuracy_score(y_test, y_pred_tuned)
precision_tuned = precision_score(y_test, y_pred_tuned)
recall_tuned = recall_score(y_test, y_pred_tuned)
f1_tuned = f1_score(y_test, y_pred_tuned)

print(f"Tuned Accuracy: {accuracy_tuned:.4f}")
print(f"Tuned Precision: {precision_tuned:.4f}")
print(f"Tuned Recall: {recall_tuned:.4f}")
print(f"Tuned F1-score: {f1_tuned:.4f}")
```

## Hyperparameter Tuning Results

GridSearchCV tested 10 combinations (C values: 0.01, 0.1, 1, 10, 100, and
solvers: liblinear, lbfgs) with 5-fold cross-validation. The best parameters were
C=1 and solver='liblinear', with a cross-validation F1-score of 0.71.

Interestingly, the tuned model's performance on the original test set decreased slightly compared to the previous (default) model—accuracy dropped from 81.01% to 78.21%, and
F1-score dropped from 0.76 to 0.72. This is because GridSearchCV selects its best model based on cross-validation performance, which doesn't always match the results on the final test set—this is a normal variation when the dataset is small. This finding is also valuable: it teaches that tuning doesn't always guarantee better results, and in the real world, it's always important to validate the model on an independent test set.
